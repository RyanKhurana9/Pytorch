```python
import torch
import torch.nn as nn

#in order to access the offerings given by the nn module the module class has to inherit it
class Model(nn.Module):
  def __init__(self,num_features):
    super().__init__()# super().__init__() is used to call the constructor (__init__) of the parent class from a child class.
    self.linear=nn.Linear(num_features,1)#(number of input features,no of output features)
    self.sigmoid=nn.Sigmoid()
  def forward(self,features):
    out=self.linear(features)#z=wx+b
    out=self.sigmoid(out)#sigmoid(z)

    #internally __call__(magic function) is over written in nn.module class=>if class Model is called=>automaticallty forward class is triggered

    return out
```

Model gets the functionalities of nn.Module because it inherits from nn.Module.
super().__init__() initializes those functionalities so they work properly.

```python
features=torch.rand(10,5)#no of features=5 (columns) 10 rows
model=Model(features.shape[1])
#model.forward(features) ->this works but not recommended by pytorch

model(features)
```
